# Analiza brendova: Smartphones,Laptops i Tablets

## O izvjestaju
U ovom izvjestaju cu analizirati performanse brendova. u tri kljucne kategorije proizvoda:**Smartphone**,**Laptops**,**Tablets**.

Cilj je da timovima omogucim brz i jasan uvid u:

-koji brendovi imaju najvece prihode,

-ko prodaje najveci broj proizvoda,

-kako korisnici ocjenjuju razlicite brendove,

-da li cijena utice na broj prodatih proizvoda.

Na kraju izvjestaja cu napraviti interaktivni HTML fajl koji timovi mogu koristiti samostalno,bez otvaranja koda.

**Autor** Adi Zebic

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.subplots as make_subplots
```

```python
df = pd.read_csv("task04_data.csv")
df.head()
```

```python
kategorije = ['Smartphone', 'Laptops', 'Tablets']
df_filtered = df[df['category'].isin(kategorije)]
print(df_filtered['category'].value_counts())
print('Ukupan broj redova:', len(df_filtered))
```

## Udio prihoda po brendu
Prikazat cu koji brendovi generisu najvise prihoda u kategorijama Smartphones,Laptops i Tablets. 

```python
prihod_po_brendu = df_filtered.groupby('brand')['revenue'].sum().reset_index()
prihod_po_brendu.columns = ['brand', 'ukupni_prihod']
prihod_po_brendu = prihod_po_brendu.sort_values("ukupni_prihod", ascending=False)

fig1 = px.pie(
    prihod_po_brendu,
    names='brand',
    values='ukupni_prihod',
    title='Udio prihoda po brendu(Smartphone, Laptops, Tablets)',
    hole=0.3
)
fig1.update_traces(textposition='inside', textinfo='percent+label')
fig1.show()
```

## Prodaja po brendu 
Ovdje cu prikazati koji brendovi prodaju najveci broj proizvoda.

```python
prodaja_po_brendu = df_filtered.groupby('brand')['quantity_sold'].sum().reset_index()
prodaja_po_brendu.columns = ['brand', 'ukupna_prodaja']
prodaja_po_brendu = prodaja_po_brendu.sort_values("ukupna_prodaja", ascending=False)

fig2 = px.bar(
    prodaja_po_brendu,
    x='brand',
    y='ukupna_prodaja',
    title='Ukupan broj prodatih proizvoda po brendu',
    labels={"brand":"Brend","ukupna_prodaja":"Broj prodatih"},
    color='brand'
)

fig2.show()
```

## Ocjene korisnika
Histogram pokazuje kako su rasporedjene ocjene,a boxplot pokazuje medijanu i raspon ocjena po brendu.

```python
fig3=px.histogram(
    df_filtered,
   x='rating',
   color='brand',
   nbins=20,
    title='Raspodjela ocjena proizvoda po brendu',
    labels={"rating":"Ocjena","count":"Broj proizvoda"},
    barmode='overlay',
    opacity=0.7
)

fig3.show()
```

```python
fig4 =px.box(
    df_filtered,
    x='brand',
    y='rating',
    title='Ocjene korisnika po brendu',
    labels={"brand":"Brend","rating":"Ocjena"}
)

fig4.show()
```

## Odnos cijene i prodaje
Preko scatter plota cu prikazati da li skuplje cijene znace manju prodaju,svaka tacka je jedan proizvod,a boja oznacava brend.

```python
fig5 = px.scatter(
    df_filtered,
    x='price',
    y='quantity_sold',
    color='brand',
    hover_name='product_name',
    title='Odnos cijene i prodaje po brendu',
    labels={"price":"Cijena","quantity_sold":"Broj prodatih"},
    size_max=15
)
fig5.show()
```

## Zakljucci
**Prihod**
Apple i Samsung generisu najveci dio ukupnog prihoda,sto ih cini kljucnim brendovima kampanje.

**Brendovi koji prodaju najvise**
Po broju prodatih proizvoda isticu se brendovi sa sirim i pristupacniji asortimanom prihoda Samsung i Apple.

**Ocjene korisnika**
Vecina brendova ima ocjene izmedju 8 i 9.5,sto je pozitivno,nema drasticnih razlika,ali neki brendovi imaju veci raspon.

**Cijena i prodaja**
Skuplji proizvodi generealno imaju manji broj prodatih komada,ali postoje izuzetci,neki skupi proizvodi se odlicno prodaju zahvaljujuci jakom brendu ili popularnom modelu.

```python
from plotly.offline import plot
import plotly.io as pio

html_sadrzaj = " "

for fig in [fig1, fig2, fig3, fig4, fig5]:
    html_sadrzaj += pio.to_html(fig, full_html=False, include_plotlyjs='cdn')

html_template = f"""
<!DOCTYPE html>
<html lang="sr">
<head>
    <meta charset="UTF-8">
    <title>Izvjestaj o brendovima</title>
    <style>
        body {{
            font-family: Arial, sans-serif;
            background-color: white;
            padding: 30px;
        }}
        h1 {{
            color: #333;
        }}
        .grafikon {{
            background:white;
            padding: 20px;
            margin-bottom: 30px;
            border-radius: 8px;
            box-shadow: 0 2px 5px rgba(0, 0, 0, 0.1);
        }}
    </style>
</head>
<body>
    <h1>Interaktivni izvjestaj:Smartphones, Laptops, Tablets</h1>
    <p>Ovaj izvještaj prikazuje performanse brendova kroz prihod, prodaju, ocjene i odnos cijene i prodaje.</p>
    
    <div class="grafikon">{html_sadrzaj}</div>
</body>
</html>
"""

with open("izvjestaj_brendovi.html", "w", encoding="utf-8") as f:
    f.write(html_template)
    
print("Izvještaj je uspješno kreiran: izvjestaj_brendovi.html")
```
